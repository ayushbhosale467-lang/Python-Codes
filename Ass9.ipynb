```python
# 1. Write a function that accepts two numbers and returns the sum, difference, using 
# multiple return values

def calculate(a,b):                                   # create a calculate function in which accept the two numbers a and b.
    return(a + b , a - b)                             # returns the vale of a+b and a-b in calculate.

num1 = int(input("Enter a first number :"))           # Take input from user.
num2 = int(input("Enter a second number :"))

sum_result, difference_result = calculate(num1, num2) # calculated values are stored in sum_result, difference_result hence we have to print these two.

print("Sum of two numbers :",sum_result)              # print the result
print("Difference of two numbers :",difference_result)
```

```python
#  2 Write a function that accepts a list of numbers and returns the largest and smallest number.

def largest_smallest(numbers):
    largest = numbers[0]
    smallest = numbers[0]

    for num in numbers :
        if num > largest :
           largest = num
        if num < smallest:
           smallest = num

    return largest , smallest
numbers = [10,20,30,40,50]

large, small = largest_smallest(numbers)

print("largest number =",large)
print("smallest number =",small)
```

```python
# Write a function using *args to calculate the sum, average, maximum, and minimum of any number of integers

def statistics(*args):
    total = sum(args)
    average = total / len(args)
    maximum = max(args)
    minimum = min(args)

    return total ,average, maximum, minimum

total, avg, maximum, minimum = statistics(10,20,30,40,50)

print("sum:",total)
print("average:",avg)
print("maximun:",maximum)
print("minimum:",minimum)
```

```python
# 4. Write a function using  **kwargs to display employee details (id, name, department, 
# salary) in a formatted manner.

def employee_details(**kwargs):
    print("Employee_Id :",kwargs["id"])

    print("Employee_Name:",kwargs["name"])

    print("Employee_Department:",kwargs ["department"])
    
    print("Employee_salary:",kwargs["salary"])

employee_details(id = "21", name = "mohan", department = "store" , salary = "20000")
employee_details(id = "88", name = "rahul", department = "food" , salary = "25000")
employee_details(id = "11", name = "suraj", department = "technical" , salary = "50000")
```

```python

#5 Write a function that accepts a string and returns the number of uppercase letters, lowercase letters, digits.


def count_characters(string):
    upper = 0 
    lower = 0
    digit = 0

    for ch in string:
        if ch.isupper():
            upper += 1 
        if ch.islower():
            lower += 1
        if ch.isdigit():
            digit += 1

    return(upper, lower, digit)

string = input("enter a string:")

upper,lower,digit = count_characters(string)
print(string)
print("numbers of uppercase :",upper)
print("numbers of lowercase :",lower)
print("numbers of digits:",digit)
```

```python
# 6 Write a function that checks whether a given string is a Palindrome

def palindrome(string):
    if string == string[::-1]:
        return True
    else:
        return False

string = input("enter a string :")

print(string)

if palindrome(string):
    print("string is palindrome.")
else:
    print("string is not palindrome.")
```

```python
# 7 Write a function that accepts a list of integers and returns a new list containing only even numbers

def even_numbers(lst):
    print("even numbers are:")
    for num in lst:
        if num % 2 == 0:
            print (num, end=" ")

numbers = [12 , 3 , 19 , 9 , 88 , 20]

even_numbers(numbers)
```

```python
# 8 Create a menu-driven calculator using functions that performs Addition, Subtraction, Multiplication, Division, and Exit.

def calculator():
    print("1. addition")
    print("2. subraction")
    print("3. multiplication")
    print("4. Division")

    choice = int(input("enter your choice:"))

    a = int(input("enter first number :"))
    b = int(input("enter second number:"))

    if choice == 1:
        print("answer =", a + b)

    elif choice == 2:
        print("answer =", a - b)

    elif choice == 3:
        print("answer =", a * b)

    elif choice == 4:
        if b != 0:
         print("answer =", a / b)
        else:
         print("division by zero is not possible.")

    else:
        print("invalid choice")

calculator()
 
    
```

```python
#9 Write a function that accepts a dictionary of students and their marks, then returns the name of the topper and the highest marks.

def top(marks):
    highest_marks = max(marks.values())

    for name in marks:
        if marks[name] == highest_marks:
            print("Topper name:", name)
            print("highest marks:", highest_marks)

top({"ayush":89 , "rahul": 74, "kunal": 65})
```

```python
#10 Write a function that accepts a sentence and returns the frequency of each word using a dictionary.

def frequency_count(string):
    frequency = {}
    for ch in string:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1
    return frequency
str1 = "Programming"
print(frequency_count(str1))
```
