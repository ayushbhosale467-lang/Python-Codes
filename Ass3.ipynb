```python
# 8. Create a simple calculator using match-case. - 
# Operations: 
# • Addition  
# • Subtraction  
# • Multiplication  
# • Division

a = int(input("enter first choice:"))
b = int(input("enter second choice:"))

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("enter the choice(1/2/3/4):"))

match choice:
    case 1:
        print("addition is =",a + b)
    case 2:
        print("subtraction is =",a - b)
    case 3:
        print("multiplication is =",a * b)
    case 4:
         if b != 0:
            print("division is =",a / b)
         else:
          print("division by zero is not allowed")
    case _:
         print("Invalid Choice")
                                  
            
```
