```python
#1 Write a program using a lambda function to find the square of each number in a list using map().

numbers = [1, 2, 3, 4, 5]

square = list(map(lambda x : x ** 2, numbers))

print("Original List :", numbers)
print("Square List :", square)
```

```python
#2 Write a program using a lambda function to convert all names in a list to uppercase using map().

names = ["ayush", "rahul", "mohan", "kartik"]

uppercase = list(map(lambda x : x.upper(), names))

print("uppercase names :", uppercase)
```

```python
# 3 Write a program using map() to calculate a 10% discount on a list of product prices.

prices = [999, 500, 654, 855, 150]

discount = list(map(lambda x : x -(x * 0.10),prices))

print("original prices :",prices)
print("after 10% discount :",discount)
```

```python
# 4 Write a program using filter() to extract all even numbers from a given list

numbers = [25, 40, 60, 18, 35, 80, 32]

even = list(filter(lambda x : x % 2 == 0, numbers))

print("Even numbers :",even)
```

```python
# 5 . Write a program using filter() to display all students who scored 35 or more marks.

marks = [25, 40, 60, 18, 35, 80, 32]

passed = list(filter(lambda x : x >= 35,marks))

print("passed students:",passed)
```

```python
#6 Write a program using filter() to remove all empty strings from a list.

word = ["Ayush", " " , "Namdev", " " , "Bhosale", " " ]

result = list(filter(lambda x : x != " " , word))

print("after removimg all strings:",result)
```

```python
#7 Write a program using map() to find the length of each string in a list.

word = ["Ayush", "namdev", "bhosale"]

result =  list(map(lambda x : len(x) , word))

print("length of each word:",result)
```

```python
# 8 . Write a program using reduce() to calculate the sum of all numbers in a list.  

from functools import reduce

numbers = [10, 20, 30, 40, 50]

total = reduce(lambda x, y : x + y, numbers)

print("sum of numbers :",total)
```

```python
#  9. Given a list of employee salaries, use map() to increase each salary by 15%.

salry = [12000, 23000, 33000, 45000, 55000]

new_salary = list(map(lambda x : x + (x * .15), salry))

print("increased salary is :", new_salary)
```

```python
# 10 Given a list of integers, use filter() to display only the numbers that are divisible by both 3 and 5.


numbers = [60, 12, 45, 2, 10, 15, 55, 33]

result = list(filter(lambda x : x % 3 == 0 and x % 5 == 0,numbers))

print("Numbers divisible by 3 and 5 are:",result)
```
