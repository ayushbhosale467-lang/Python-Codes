```python
# Electricity Bill Calculation

# Variables
units_consumed = 300
cost_per_unit = 8

# Calculate bill
total_bill = units_consumed * cost_per_unit
gst = total_bill * 0.18
final_bill = total_bill + gst

# Display results
print("Units Consumed:", units_consumed)
print("Cost per Unit: ₹", cost_per_unit)
print("Total Bill (Without GST): ₹", total_bill)
print("GST (18%): ₹", gst)
print("Final Bill (Including GST): ₹", final_bill)
```

```python
# Create a list of programming languages
languages = ["Python", "C++", "JavaScript", "C", "PHP"]

# Membership Operators
print("Python exists:", "Python" in languages)
print("Java does not exist:", "Java" not in languages)
print("C++ exists:", "C++" in languages)
```

```python
# int → float
num = 25
float_num = float(num)
print("int → float:", float_num)

# float → string
price = 99.5
string_price = str(price)
print("float → string:", string_price)

# string → int
age = "20"
int_age = int(age)
print("string → int:", int_age)

# int → bool
number = 1
bool_value = bool(number)
print("int → bool:", bool_value)

# bool → int
is_pass = True
int_value = int(is_pass)
print("bool → int:", int_value)
```

```python
# Product Information

# Store product details
product_name = "Laptop"
price = 50000
quantity = 2

# Calculate total cost
total_cost = price * quantity

# Calculate GST (18%)
gst = total_cost * 0.18

# Calculate final amount
final_amount = total_cost + gst

# Display results
print("Product Name:", product_name)
print("Price: ₹", price)
print("Quantity:", quantity)
print("Total Cost: ₹", total_cost)
print("GST (18%): ₹", gst)
print("Final Amount: ₹", final_amount)
```
