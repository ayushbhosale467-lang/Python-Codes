```python
# 1. Write a program to count the frequency of each character in a string using a 
# dictionary. - - -  - - 
# Input: 
# "programming" 
# Output: 
# {'p':1, 'r':2, 'o':1, ...} 

str1 = "PROGRAMMING"

freq = {}

for i in str1:
    freq[i] = freq.get(i , 0) + 1

print("frequency of each character:")
print("-----------------------")
print(freq)
```

```python
# 2. Write a program to count the frequency of each element in a list using a dictionary. - - -  - - 
# [10,20,10,30,20,10] 
# Output: 
# {10:3, 20:2, 30:1} 

listone = [10,20,10,30,20,10]
freq = {}

for i in listone:
    if i in freq:
        freq[i] += 1 
    else:
        freq[i] = 1

print ("frequency of each element",freq)
```

```python
# 3. Write a program to find the key having the maximum value in a dictionary. - - -  - - 
# {'A':85, 'B':92, 'C':78} 
# Output: 
# B 

marks =  {'A':85, 'B':92, 'C':78}

max_key = None
max_value = 0

for i,j in marks.items():
    if j > max_value:
        max_value = j
        max_key = i

print("maximun value in dictionary ", max_key)
```

```python
# 4. Write a program to find the key having the minimum value in a dictionary.

dict1 = {'A': 85, 'B': 92, 'C': 78}

min_key = ""
min_value = 100

for key in dict1:
    if dict1[key] < min_value:
        min_value = dict1[key]
        min_key = key

print("the key having minimum value in ",min_key)
```

```python
# 6. Write a program to remove duplicate values from a dictionary. - - -  - - 
# {'a':10, 'b':20, 'c':10, 'd':30} 
# Output: 
# {'a':10, 'b':20, 'd':30} 

dict1 = {'a':10, 'b':20, 'c':10, 'd':30}
result = {}
seen = set()

for i,j in dict1.items():
    if j not in seen:
        result[i]=j
        seen.add(j)
print(result)
```

```python
# 8. Write a program to find the sum of all values in a dictionary. - 
# {'Math':85, 'Science':90, 'English':80} -  - - 
# Output: 
# 255

dict1 = {'Math':85, 'Science':90, 'English':80} 
sum = 0 
for  i in dict1:
    sum = sum + dict1[i]
print(sum)
```

```python
# 9. Write a program to create a dictionary from two lists: one containing keys and the 
# other containing values. - - 
# Keys = ['A', 'B', 'C'] 
# Values = [10, 20, 30]

keys = ['A', 'B', 'C']
values = [10, 20, 30]
result = {}

for i in keys:
    for j in values:
     result[i] = j
print(result)
```

```python
# 10. Write a program to create a student result dictionary containing name and marks of 
# 5 students, then display the topper. 

student = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    student[name] = marks

topper = ""
highest = 0

for name in student:
    if student[name] > highest:
        highest = student[name]
        topper = name

print("Topper:", topper)
print("Marks:", highest)
```
