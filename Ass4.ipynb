```python
# 1. Write a program to count the number of vowels and consonants in a given string. 

string = input("enter your string :").lower()

v = 0
c = 0

for i in string :
    if i in "aeiou":
        v += 1 
    elif i.isalpha():
        c += 1     

print("vowels:",v)
print("consonats:",c)
```

```python
# 2. Write a program to check whether a given string is a Palindrome. - - 
# Input: madam 
# Output : Palindrom

string = input("Enter your string :")

temp = "".join(reversed(string))

if string == temp :
   print("it is a palidrome.")
else:
    print("it is not palindrome.")
```

```python
# 3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions.

string = "ayush"

reversed = ""

for i in string:
    reversed = i + reversed

print("reversed string is :",reversed)
```

```python
# 5. Write a program to remove all duplicate characters from a string while maintaining their 
# original order. - - 
# Input : programming 
# Output : progamin


string = "programming"
result = ""

for i in string:
    if i not in result:
      result += i

print("output:",result)
```

```python
# 6. Write a program to find the longest word in a sentence. - 
# Input: Python is an amazing programming language  - 
# Output: programming 

string = " Python is an amazing programming language."

words = string.split()       # ["Python", "is", "an", "amazing", "programming", "language"]     
longest = ""

for i in words:
    if len(i) > len(longest):   # If the current word is longer than the previous longest word, update longest.
        longest = i
print("Longest word is :",longest)
```

```python
# 7. Write a program to capitalize the first letter of every word without using the title() 
# method. - - 
# Input:    welcome to python 
# Output:   Welcome To Python

string = "welcome to python"

words = string.split()

for i in words:
   print(i[0].upper() + i [1:],end=" ")
```

```python
# 8. Write a program to count the total number of words in a sentence without using the 
# split() method. - - 
# Input:     Python is easy to learn 
# Output:    5 

sentence = "python is easy to learn"

count = 1 

for i in sentence:
    if i == " ":
       count += 1
print("total words =",count)
```
