```python
#1 Write a program to create a text file and store the name, age, and marks of a student in it

name = input("enter your name:")
age = int(input("enter your age:"))
marks = int(input("enter your marks"))

with open ("temp.txt","w")as file:
    file.write("name:"+ name + "\n")
    file.write("age:"+ str(age) + "\n")
    file.write("marks:"+ str(marks) + "\n")

file.close()

print("student details stored successfully:")
```

```python
# 9 Write a program to read a file containing numbers and calculate their sum, average, maximum, and minimum.


with open ("numbers.txt", "w") as file1:
    file1.write("10\n")
    file1.write("20\n")
    file1.write("30\n")
    file1.write("40\n")
    file1.write("50\n")
    file1.write("60\n")
    file1.write("70\n")
    file1.write("80\n")
    file1.write("90")

with open ("numbers.txt", "r") as file2:

    numbers = []

    for line in file2:
        numbers.append(int(line.strip()))

total = sum(numbers)
average = total / len(numbers)
maximum = max(numbers)
minimum = min(numbers)


print("sum of numbers :",total)
print("average:", average)
print("maximum:",maximum)
print("minimum:", minimum)
```
