```python
#1 Write a program to create a text file and store the name, age, and marks of a student in it

name = input("enter your name:")
age = int(input("enter your age:"))
marks = int(input("enter your marks"))

with open ("temp.txt","w")as file:
    file.write("name:"+ name + "\n")
    file.write("age:"+ str(age) + "\n")
    file.write("marks:"+ str(marks) + "\n")

file.close()

print("student details stored successfully:")
```

```python
#2  Write a program to read the complete contents of a file and count the total number of characters.

with open ("temp.txt", "r")as file:
    data  = file.read()

print("file contents :")
print(data)

print("total numbers of characters:",len(data))

```

```python
 # 3 Write a program to read a file line by line and count the total number of lines. 

with open ("temp.txt", "r") as file:
    lines = file.readlines()
print("lines in files :")
print(lines)
print("total numbers of lines:",len(lines))
```
