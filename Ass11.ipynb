Python Exception Handling 


```python
# 1. Write a program to handle ZeroDivisionError when dividing two numbers entered by the user.


try:
    num1 = int(input("enter a first choice:"))
    num2 = int(input("enter a second choice:"))

    result = num1 / num2

except ZeroDivisionError as err:
    print("you can not divide by zero.",err)

else:
    print("result is:",result)

finally:
    print("ZeroDivisionError is solved successfully ..!")
```

```python
# 2. Write a program to handle ValueError when the user enters a non-numeric value for age.  

try:
    age = int(input("enter your age:"))

except ValueError as err:
    print("please enter your age in numeric format.",err)

else:
    print("your age is:",age)
```

```python
#3. Write a program to handle IndexError while accessing an element from a list.

numbers = [10, 20 ,30 , 40 ,50]

try:
    index = int(input("enter index:"))
    value = numbers[index]

except IndexError as err:
    print("index is not out of range.",err)

else:
    print("value of index you entered is:",value)

finally:
    print("indexerror handling is completed.")    
```

```python
# 4. Write a program to handle KeyError when accessing a key from a dictionary.

student = {"name" : "Ayush", "age" : 22, "course" : "cs"}

try:
    key = input("Enter a key :")
    value = student[key]

except KeyError as err:
    print("Key does not exist in dictionary.",err)

else:
    print("Value of the key you entered is :",value)
```

```python
# 5. Write a program to handle TypeError by attempting to add an integer and a string.  

try:
    num1 = int(input("Enter a first number :"))
    num2 = input("Enter a second number :")

    result = num1 + num2

    print("First number is :",num1)
    print("Second number is :",num2)

except TypeError as err:
    print("Please entered integer value.",err)

else:
    print("Result is :",result)
```

```python
# 6 Write a program that demonstrates the use of try, except, else, and finally blocks.

try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

except ZeroDivisionError as err:
    print("Error: Cannot divide by zero.")

except ValueError as err:
    print("Error: Please enter valid numbers.")

except TypeError as err:
    print("Please entered integer value.",err)

else:
    print("Division successful.")
    print("Result:", result)

finally:
    print("Program executes successfully.")
```
