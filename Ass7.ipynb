```python
# 1. Write a program to find the union of two sets without using the union() method. 

set1 = {11,15,45,85,66}
set2 = {66,86,45,71,22}

union = set1.copy()

for i in set2:
    union.add(i)

print("union of two sets are:",union)
```

```python
# 2. Write a program to remove all duplicate elements from a list using a set.

listone = (1,2,3,2,4,5,6,1,7,3,9,8,6,9)

unique = list(set(listone))

print("remove all duplicates:",unique)

```

```python
# Write a program to find all common elements between two lists using sets. 

listone = {11,15,45,85,66}
listtwo = {66,86,45,71,22}

setone = set(listone)
settwo = set(listtwo)

common = setone.intersection(settwo)

print("common elements  are :",common)

```

```python
# 4. Write a program to count the number of unique elements in a list using a set.

numbers = [1,2,3,4,5,6,5,4,3,7,8,9,10,11,12]

unique = set(numbers)

print("unique numbers",list(unique))
print("count of numbers",len(unique))
```

```python
# 5. Write a program to find the missing numbers between 1 to N using sets. - - - -  - - 
# Input: 
# N = 10 
# Numbers = {1,2,3,5,7,8,10} 
# Output: 
# {4,6,9} 

n = 10
numbers = {1,2,3,5,7,8,10}

all_numbers = set(range(1,11))

missing = all_numbers - numbers

print("Missing numbers are :",missing)
```

```python
# 6. Write a program to create a new set containing only the even numbers from a given 
# set. 

numbers = {1,2,4,5,7,8,14,17,20,22,50,101}

even = set()

for i in numbers:
    if i % 2 == 0:
        even.add(i)

print("Even numbers from given set are :",even)
```

```python
# 7. Write a program to merge three sets into a single set containing only unique 
# elements

set1 = {1,2,3,4,5,6,7,8}
set2 = {5,6,7,8,9,10,11}
set3 = {9,10,11,12,13,14}

merge = set1.copy()

for i in set2:
    merge.add(i)

    for i in set3:
        merge.add(i)

print("unique elements after merging 3 sets:",merge)
```

```python
# 8. Given two strings, find the common characters using sets.

str1 = "ayush"
str2 = "bhosale"

set1 = set(str1.upper())
set2 = set(str2.upper())

common = set1.intersection(set2)

print("common characters :",common)
```
