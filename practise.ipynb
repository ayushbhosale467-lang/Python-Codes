```python
# finding the midlle element 

listone = [1,2,3]
listtwo = [4,5,6]

result = listone + listtwo
print(result)

if len(result)% 2 == 0:
    indexmum = len(result)//2

    num1 = result[indexmum]
    num2 = result[indexmum-1]

print((num1 + num2)/2)

```

```python
# find the occurance of numbers and which occurs for more times

list1 = [3, 2, 3]

result = {}

for i in list1:
    if i in result:
        result[i] += 1
    else:
        result[i] = 1 

highest = max(result, key=result.get)

print("highest occurance element:",highest)
print("count:", result[highest])

 
```

```python
# count the highest number and its occurence

numbers = [3, 1, 3, 2, 1, 2, 4]

highest = 0 
highest_count = 0


for i in numbers:
    count = numbers.count(i)

    if count > highest_count:
        highest_count = count
        highest = i

print("occurance:",highest_count)
print("highest:",highest)

```
